```python
import ast
```

```python
def extract_names(text):
    items = ast.literal_eval(text)

    return [
        item["name"]
        for item in items
    ]
```

```python
movies['genres'] = movies['genres'].apply(extract_names)
```

```python
movies['genres']
```

```python
movies['keywords'] = movies['keywords'].apply(extract_names)
```

```python
movies['keywords'].iloc[0]
```

```python
def extract_top_casts(text):
    items = ast.literal_eval(text)

    return [
        item['name']
        for item in items[:3]
    ]
```

```python
movies['cast'] = movies['cast'].apply(extract_top_casts)
```

```python
movies['cast']
```

```python
movies['crew'].iloc[0]
```

```python
def extract_director(text):
    items = ast.literal_eval(text)

    for item in items:
        if item['job'] == 'Director':
            return item['name']
    return ""
```

```python
movies['crew'] = movies['crew'].apply(extract_director)
```

```python
movies['crew'].iloc[0]
```

```python
movies = movies.rename(
    columns={'crew' : 'director'}
)
```

```python
movies['director'] = movies['director'].astype(object)
movies['original_title'] = movies['original_title'].astype(object)
movies['overview'] = movies['overview'].astype(object)
```

```python
movies = pd.read_csv("../data/movies_clean_data.csv")
```

```python
movies.info()
```

```python
movies["genres"] = movies["genres"].apply(ast.literal_eval)
movies["keywords"] = movies["keywords"].apply(ast.literal_eval)
movies["cast"] = movies["cast"].apply(ast.literal_eval)
```

```python
movies.head(2)
```

```python
def normalize_list(items):
    return [item.lower() for item in items]
```

```python
movies["genres"] = movies["genres"].apply(normalize_list)
movies["keywords"] = movies["keywords"].apply(normalize_list)
movies["cast"] = movies["cast"].apply(normalize_list)
```

```python
movies["director"] = movies["director"].fillna("")
movies["director"] = movies["director"].str.lower()
```

```python
movies["overview"] = movies["overview"].fillna("")
movies["overview"] = movies["overview"].str.lower()
```

```python
def join_list(items):
    return " ".join(items)
```

```python
movies["genres"] = movies["genres"].apply(join_list)
movies["keywords"] = movies["keywords"].apply(join_list)
movies["cast"] = movies["cast"].apply(join_list)
```

```python
movies[["original_title", "genres", "keywords", "cast", "director", "overview"]].head()
```

```python
# Putting every feature into a single "tag" column, to represent then as a text document to "TF-IDF".
movies["tags"] = (
    movies["genres"] + " " +
    movies["keywords"] + " " +
    movies["cast"] + " " +
    movies["director"] + " " +
    movies["overview"]
)
```

```python
movies["tags"].str.len().describe()
```

```python
movies = movies[
    ["id", "original_title", "tags"]
]
```

```python
movies.head(3)
```

```python
import re
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()
```

```python
movies["tags"] = movies["tags"].apply(clean_text)
```

```python
movies.loc[
    movies["original_title"] == "Interstellar",
    ["original_title", "tags"]
]
```
